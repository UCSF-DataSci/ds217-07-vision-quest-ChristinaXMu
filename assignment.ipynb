### Part 2.2: Statistical Plots

```python
# TODO: Create a figure with 2x2 subplots
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# TODO: Top-left: Box plot of total_amount by store_location
# TODO: Top-right: Violin plot of total_amount by category
# TODO: Bottom-left: Scatter plot of quantity vs total_amount with hue by gender
# TODO: Bottom-right: Histogram of total_amount with kde overlay

# TODO: Add appropriate titles and labels
# TODO: Customize colors and styling
# TODO: Save the plot as 'output/q2_seaborn_plots.png'

# ============================================================================
# QUESTION 2: seaborn Statistical Visualization
# ============================================================================
print("="*60)
print("QUESTION 2: seaborn Statistical Visualization")
print("="*60)

# Merge datasets for seaborn analysis
merged_df = sales.merge(products, on='product_id').merge(customers, on='customer_id')
print("\n✓ Datasets merged successfully")
print(f"Merged data shape: {merged_df.shape}")

# Set seaborn style
sns.set_style("whitegrid")
sns.set_palette("Set2")

# ============================================================================
# TASK 1: Create 2x2 seaborn statistical plots
# ============================================================================
print("\nCreating seaborn statistical plots...")

# Create a figure with 2x2 subplots
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
fig.suptitle('seaborn Statistical Visualizations', fontsize=16, fontweight='bold')

# Top-left: Box plot of total_amount by store_location
sns.boxplot(data=merged_df, x='store_location', y='total_amount', 
            ax=axes[0, 0], palette='Set2', width=0.6)
axes[0, 0].set_title('Sales Distribution by Store Location', fontsize=12, fontweight='bold')
axes[0, 0].set_xlabel('Store Location', fontsize=11)
axes[0, 0].set_ylabel('Total Amount ($)', fontsize=11)
axes[0, 0].grid(True, alpha=0.3, axis='y')

# Top-right: Violin plot of total_amount by category
sns.violinplot(data=merged_df, x='category', y='total_amount', 
               ax=axes[0, 1], palette='muted', inner='quartile')
axes[0, 1].set_title('Sales Distribution by Product Category', fontsize=12, fontweight='bold')
axes[0, 1].set_xlabel('Category', fontsize=11)
axes[0, 1].set_ylabel('Total Amount ($)', fontsize=11)
axes[0, 1].tick_params(axis='x', rotation=45)
plt.setp(axes[0, 1].xaxis.get_majorticklabels(), rotation=45, ha='right')

# Bottom-left: Scatter plot of quantity vs total_amount with hue by gender
sns.scatterplot(data=merged_df, x='quantity', y='total_amount', 
                hue='gender', ax=axes[1, 0], palette='coolwarm', 
                s=80, alpha=0.6, edgecolor='black', linewidth=0.5)
axes[1, 0].set_title('Quantity vs Total Amount by Gender', fontsize=12, fontweight='bold')
axes[1, 0].set_xlabel('Quantity', fontsize=11)
axes[1, 0].set_ylabel('Total Amount ($)', fontsize=11)
axes[1, 0].legend(title='Gender', fontsize=10, title_fontsize=11)
axes[1, 0].grid(True, alpha=0.3)

# Bottom-right: Histogram of total_amount with kde overlay
sns.histplot(data=merged_df, x='total_amount', bins=30, 
             kde=True, ax=axes[1, 1], color='steelblue', 
             edgecolor='black', linewidth=1, alpha=0.7)
axes[1, 1].set_title('Distribution of Transaction Amounts', fontsize=12, fontweight='bold')
axes[1, 1].set_xlabel('Total Amount ($)', fontsize=11)
axes[1, 1].set_ylabel('Frequency', fontsize=11)
axes[1, 1].grid(True, alpha=0.3, axis='y')

# Use tight layout
plt.tight_layout()

# Save the plot as 'output/q2_seaborn_plots.png'
plt.savefig('output/q2_seaborn_plots.png', dpi=300, bbox_inches='tight')
print("✓ Saved: output/q2_seaborn_plots.png")
plt.show()
plt.close()
```

```python
# TODO: Convert transaction_date to datetime
# merged_df['transaction_date'] = pd.to_datetime(merged_df['transaction_date'])

# TODO: Create daily sales aggregation
# daily_sales = merged_df.groupby('transaction_date')['total_amount'].sum()

# TODO: Create a figure with 2x2 subplots
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# TODO: Top-left: Line plot of daily sales
# TODO: Top-right: Rolling 7-day average
# TODO: Bottom-left: Monthly sales bar chart (use resample('ME') for month-end frequency)
# TODO: Bottom-right: Sales distribution histogram

# TODO: Add titles and labels
# TODO: Save the plot as 'output/q3_pandas_plots.png'

# ============================================================================
# QUESTION 3: pandas Plotting and Data Exploration
# ============================================================================
print("\n" + "="*60)
print("QUESTION 3: pandas Plotting and Data Exploration")
print("="*60)

# Ensure transaction_date is datetime (already done above, but good practice)
merged_df['transaction_date'] = pd.to_datetime(merged_df['transaction_date'])

# Create daily sales aggregation
daily_sales = merged_df.groupby('transaction_date')['total_amount'].sum()
print(f"\n✓ Daily sales aggregated: {len(daily_sales)} days")

# ============================================================================
# TASK 1: Time series analysis with pandas plotting
# ============================================================================
print("\nCreating pandas time series plots...")

# Create a figure with 2x2 subplots
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
fig.suptitle('pandas Time Series and Distribution Analysis', fontsize=16, fontweight='bold')

# Top-left: Line plot of daily sales
daily_sales.plot(ax=axes[0, 0], color='steelblue', linewidth=2, marker='o', markersize=4)
axes[0, 0].set_title('Daily Sales Trend', fontsize=12, fontweight='bold')
axes[0, 0].set_xlabel('Date', fontsize=11)
axes[0, 0].set_ylabel('Total Sales ($)', fontsize=11)
axes[0, 0].grid(True, alpha=0.3)
axes[0, 0].tick_params(axis='x', rotation=45)

# Top-right: Rolling 7-day average
rolling_avg = daily_sales.rolling(window=7).mean()
daily_sales.plot(ax=axes[0, 1], color='lightblue', linewidth=1.5, alpha=0.5, label='Daily Sales')
rolling_avg.plot(ax=axes[0, 1], color='red', linewidth=2.5, label='7-Day Moving Average')
axes[0, 1].set_title('Sales with 7-Day Moving Average', fontsize=12, fontweight='bold')
axes[0, 1].set_xlabel('Date', fontsize=11)
axes[0, 1].set_ylabel('Total Sales ($)', fontsize=11)
axes[0, 1].legend(fontsize=10)
axes[0, 1].grid(True, alpha=0.3)
axes[0, 1].tick_params(axis='x', rotation=45)

# Bottom-left: Monthly sales bar chart (use resample('ME') for month-end frequency)
monthly_sales = daily_sales.resample('ME').sum()
monthly_sales.plot(kind='bar', ax=axes[1, 0], color='coral', edgecolor='black', width=0.7)
axes[1, 0].set_title('Monthly Sales', fontsize=12, fontweight='bold')
axes[1, 0].set_xlabel('Month', fontsize=11)
axes[1, 0].set_ylabel('Total Sales ($)', fontsize=11)
axes[1, 0].tick_params(axis='x', rotation=45)
axes[1, 0].grid(True, alpha=0.3, axis='y')
# Format x-axis labels to show month-year
axes[1, 0].set_xticklabels([date.strftime('%b %Y') for date in monthly_sales.index], rotation=45, ha='right')

# Bottom-right: Sales distribution histogram
merged_df['total_amount'].plot(kind='hist', bins=30, ax=axes[1, 1], 
                                 color='mediumseagreen', edgecolor='black', alpha=0.7)
axes[1, 1].set_title('Sales Distribution Histogram', fontsize=12, fontweight='bold')
axes[1, 1].set_xlabel('Total Amount ($)', fontsize=11)
axes[1, 1].set_ylabel('Frequency', fontsize=11)
axes[1, 1].grid(True, alpha=0.3, axis='y')

# Use tight layout
plt.tight_layout()

# Save the plot as 'output/q3_pandas_plots.png'
plt.savefig('output/q3_pandas_plots.png', dpi=300, bbox_inches='tight')
print("✓ Saved: output/q3_pandas_plots.png")
plt.show()
plt.close()
```
